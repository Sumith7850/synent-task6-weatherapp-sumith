```python
import requests

api_key = "3ed8ee782ade43589e545136262104 "

cities = input("Enter cities (comma separated): ").split(",")

print("\nFetching weather data...\n")

for city in cities:
    city = city.strip()

    url = f"http://api.weatherapi.com/v1/current.json?key={api_key}&q={city}"

    response = requests.get(url)
    data = response.json()

    if "error" not in data:
        temp = data["current"]["temp_c"]
        humidity = data["current"]["humidity"]
        weather = data["current"]["condition"]["text"]

        print(f"City: {city}")
        print(f"Temperature: {temp}°C")
        print(f"Humidity: {humidity}%")
        print(f"Weather: {weather}")
        print("-" * 25)
    else:
        print(f"City: {city} not found!")
```

```python
import csv

with open("weather_data.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["City", "Temperature", "Humidity", "Weather"])
```
